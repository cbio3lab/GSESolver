```python
from calculateDescriptors import calculateDescriptors
import pandas as pd

def create_descriptors_table(smiles,filename="descriptors.csv"):
    if len(smiles) == 0:
      return
    unique_smiles = len(smiles) == 1
    if unique_smiles:
      smiles = smiles + ["C"]
    descriptors = calculateDescriptors(smiles)
    if unique_smiles:
      descriptors.drop(descriptors.tail(1).index,inplace = True)
      smiles = smiles[:-1]
    descriptors.to_csv(filename,index=False)
    print(f"{filename} created")
```

```python
fileName = "APPLICATIONS.xlsx"

#read the SMILES codes
fileFound = False
try:
  df = pd.read_excel(fileName)
  fileFound = True
except FileNotFoundError:
  print("File not found")

smiles_col_name = None
for col in df.columns:
  if col.upper() == "SMILES":
    smiles_col_name = col
    break

if smiles_col_name is None:
    raise KeyError("Column 'SMILES' not found in the Excel file.")

descriptors = create_descriptors_table(df[smiles_col_name])
```
