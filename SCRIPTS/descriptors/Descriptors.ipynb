```python
from calculateDescriptors import calculateDescriptors
import pandas as pd

def create_descriptors_table(smiles,filename="descriptors.csv"):
    if len(smiles) == 0:
      return
    unique_smiles = len(smiles) == 1
    if unique_smiles:
      smiles = smiles + ["C"]
    descriptors = calculateDescriptors(smiles)
    if unique_smiles:
      descriptors.drop(descriptors.tail(1).index,inplace = True)
      smiles = smiles[:-1]
    descriptors.to_csv(filename,index=False)
    print(f"{filename} created")
```

```python
filename = "SolCbio3Database_merged.csv"
fileFound = False
try:
  df = pd.read_csv(filename)
  fileFound = True
except FileNotFoundError:
  print("File not found")

if fileFound and "smiles" in df.columns:
    smiles = pd.read_csv(filename)["smiles"]
elif fileFound and "SMILES" in df.columns:
    smiles = pd.read_csv(filename)["SMILES"]
elif fileFound and "Smiles" in df.columns:
    smiles = pd.read_csv(filename)["Smiles"]
else:
    print("SMILES not found")
    smiles = []
create_descriptors_table(smiles,f"descriptors_{filename}")
```
